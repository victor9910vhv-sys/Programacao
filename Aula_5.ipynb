Exercício 1 — Contagem

Faça um programa que peça um número ao usuário e mostre todos os números de 1 até esse número usando while

```python
n = int(input("Digite um número: "))
i = 1
while i <= n:
    print(i)
    i += 1
```

Exercício 2 — Contagem regressiva

Peça um número ao usuário e faça uma contagem até 0, mostrando cada número na tela.

```python
n = int(input("Digite um número: "))
while n >= 0:
    print(n)
    n -= 1
```

Exercício 3 — Números pares

Faça um programa que mostre todos os números pares de 0 até 20 usando while.

```python
i = 0
while i <= 20:
    print(i)
    i += 2
```

Exercício 4 — Soma dos números

Peça um número ao usuário e calcule a soma de todos os números de 1 até esse número.

```python
n = int(input("Digite um número: "))
soma = 0
i = 1
while i <= n:
    soma += i
    i += 1
print("Soma:", soma)
```

Exercício 5 — Tabuada

Peça um número ao usuário e mostre a tabuada desse número de 1 até 10 usando while.

```python
n = int(input("Digite um número: "))
i = 1
while i <= 10:
    print(n, "x", i, "=", n * i)
    i += 1
```

Exercício 6 — Senha

Crie um programa que peça uma senha ao usuário.

Enquanto a senha estiver incorreta, mostre:

Senha incorreta!

Quando o usuário digitar a senha correta, mostre:

Acesso permitido!

Considere que a senha correta seja "1234".

```python
senha = ""
while senha != "1234":
    senha = input("Digite a senha: ")
    if senha != "1234":
        print("Senha incorreta!")
print("Acesso permitido!")
```

Exercício 7 — Somando até digitar zero

Faça um programa que peça vários números ao usuário.

O programa deve continuar pedindo números até que o usuário digite 0.

Ao final, mostre a soma de todos os números digitados.

```python
soma = 0
n = 1  # começa com 1 apenas para entrar no loop
while n != 0:
    n = int(input("Digite um número: "))
    soma += n
print("Soma:", soma)
```

Exercício 8 — Maior número

Peça números ao usuário até que ele digite 0.

Ao final, mostre qual foi o maior número digitado.

```python
maior = 0
n = 1
while n != 0:
    n = int(input("Digite um número: "))
    if n > maior:
        maior = n
print("Maior número:", maior)
```

Exercício 9 — Média das notas

Faça um programa que peça 5 notas de um aluno usando while.

Ao final, calcule e mostre a média das notas.

```python
soma = 0
i = 1
while i <= 5:
    soma += float(input("Digite a nota: "))
    i += 1
print("Média:", soma / 5)
```

Exercício 10 — Caixa eletrônico

Faça um programa que comece com um saldo de R$ 1000,00.

Mostre um menu:

1 - Depositar

2 - Sacar

3 - Ver saldo

4 - Sair

```python
saldo = 1000.0
opcao = 0
while opcao != 4:
    print("\n1-Depositar | 2-Sacar | 3-Saldo | 4-Sair")
    opcao = int(input("Opção: "))
    if opcao == 1:
        saldo += float(input("Valor: R$ "))
    elif opcao == 2:
        saldo -= float(input("Valor: R$ "))
    elif opcao == 3:
        print("Saldo: R$", saldo)
```
