LISTAS

```python
# 1. Crie uma lista chamada frutas contendo: maçã, banana, laranja e morango. Mostre a lista.

frutas = ["maçã", "banana", "laranja", "morango"]

print(frutas)
```

```python
# 2. Crie a lista nomes = ["Rebeca", "Guilherme", "Melissa", "Kyara"]. Mostre o primeiro, o segundo e o último nome.

nomes = ["Rebeca", "Guilherme", "Melissa", "Kyara"]

print(nomes[0])
print(nomes[1])
print(nomes[-1])
```

```python
# 3. Crie cores = ["vermelho", "azul", "verde"]. Troque "azul" por "amarelo" e mostre a lista.

cores = ["vermelho", "azul", "verde"]

cores[1] = "amarelo"

print(cores)
```

```python
# 4. Crie alunos = ["Nicolas", "Wendel", "Pyetra"]. Adicione "Petros" usando append().

alunos = ["Nicolas", "Wendel", "Pyetra"]

alunos.append("Petros")

print(alunos)
```

```python
# 5. Crie animais = ["cachorro", "gato", "coelho", "passarinho"]. Remova "coelho" usando remove().

animais = ["cachorro", "gato", "coelho", "passarinho"]

animais.remove("coelho")

print(animais)
```

TUPLAS

```python
# 1. Crie uma tupla chamada meses contendo Janeiro, Fevereiro, Março e Abril. Mostre a tupla.

meses = ("Janeiro", "Fevereiro", "Março", "Abril")

print(meses)
```

```python
# 2. Crie numeros = (10, 20, 30, 40, 50). Mostre o primeiro, o terceiro e o último número.

numeros = (10, 20, 30, 40, 50)

print(numeros[0])
print(numeros[2])
print(numeros[-1])
```

```python
# 3. Crie uma tupla aluno contendo o nome, a idade e o curso de um aluno. Mostre cada informação separadamente.

aluno = ("Victor", 16, "Informática")

print(aluno[0])
print(aluno[1])
print(aluno[2])
```

```python
# 4. Crie cores = ("vermelho", "azul", "verde"). Tente alterar "azul" para "amarelo". Observe o que acontece.

cores = ("vermelho", "azul", "verde")

cores[1] = "amarelo"

print(cores)
```

```python
# 5. Crie numeros = (2, 4, 2, 6, 2, 8, 10). Use count() para descobrir quantas vezes o número 2 aparece.

numeros = (2, 4, 2, 6, 2, 8, 10)

print(numeros.count(2))
```
