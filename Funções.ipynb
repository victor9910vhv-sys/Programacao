```python
# 1. Olá, mundo!

def ola():
    print("Olá, mundo!")

ola()
```

```python
# 2. Mensagem

def mensagem():
    print("Bem-vindo ao curso de programação!")

mensagem()
mensagem()
```

```python
# 3. Soma

def somar():
    print(10 + 20)

somar()
```

```python
# 4. Subtração

def subtrair():
    print(50 - 15)

subtrair()
```

```python

# 5. Multiplicação

def multiplicar():
    print(8 * 7)

multiplicar()
```

```python
# 6. Cumprimentar

def cumprimentar(nome):
    print("Olá,", nome + "!")

cumprimentar("Maria")
cumprimentar("João")
cumprimentar("Carlos")
```

```python
# 7. Dobro

def dobro(numero):
    print("Número:", numero)
    print("Dobro:", numero * 2)

dobro(5)
```

```python
# 8. Idade

def mostrar_idade(idade):
    print("Você tem", idade, "anos.")

mostrar_idade(18)
mostrar_idade(20)
mostrar_idade(16)
```

```python
# 9. Positivo ou negativo

def verificar_numero(numero):
    if numero > 0:
        print("O número é positivo.")
    elif numero < 0:
        print("O número é negativo.")
    else:
        print("O número é zero.")

verificar_numero(10)
verificar_numero(-5)
verificar_numero(0)
```

```python
# 10. Média

def calcular_media(nota1, nota2):
    media = (nota1 + nota2) / 2
    print("Nota 1:", nota1)
    print("Nota 2:", nota2)
    print("Média:", media)

calcular_media(8, 6)
```
