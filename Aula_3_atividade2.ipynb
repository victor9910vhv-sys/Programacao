Exercício 10 – Calculadora de IMC
Receba o peso e a altura de uma pessoa, calcule o IMC usando a fórmula:
IMC = peso / (altura × altura)
Depois, classifique o resultado:
 IMC menor que 18.5 → Abaixo do peso
 IMC de 18.5 a 24.9 → Peso normal
 IMC de 25 a 29.9 → Sobrepeso
 IMC maior ou igual a 30 → Obesidade

```python
p= float(input("qual seu peso? "))
a= float(input("qual sua altura? "))

imc= p / (a * a)

if imc < 18.5 and imc > 15 :
  print("Você está a baixo do peso")

elif imc >= 18.5 and imc <= 24.9:
  print("Você está com o peso normal")

elif imc >= 25 and imc <= 29.9:
  print("Você está com sobrepeso")

elif imc >= 30:
  print("Você está com obesidade")

else:
  print("Como você ainda está vivo?!")
```
