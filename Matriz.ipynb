# 1. Matriz 2x2

```python
matriz = [[1, 2], [3, 4]]
print(matriz[0])
print(matriz[1])
```

# 2. Matriz 2x3

```python
matriz2 = [[1, 2, 3], [4, 5, 6]]
print(matriz2[0])
print(matriz2[1])
```

# 3. Matriz com nomes

```python
matriz_nomes = [["Ana", "João"], ["Maria", "Pedro"]]
print(matriz_nomes[0])
print(matriz_nomes[1])
```

# 4. Matriz com números de 1 a 9

```python
matriz3 = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(matriz3[0])
print(matriz3[1])
print(matriz3[2])
```

# 5. Matriz 3x3 com zeros

```python
matriz_zeros = [[0, 0, 0], [0, 0, 0], [0, 0, 0]]
print(matriz_zeros[0])
print(matriz_zeros[1])
print(matriz_zeros[2])
```
