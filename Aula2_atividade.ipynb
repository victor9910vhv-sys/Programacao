Exercício 6 – Situação prática
Um produto custa R$ 80,00. O cliente ganhou 10% de desconto.
Faça um programa que calcule e mostre:
• O valor do desconto
• O valor final a pagar
• Se o valor final é menor que 75.00 (True ou False)

```python
n= 80.00
d= 10

vd =n * (d / 100)
f = n - vd
l = f < 75.00

print(f"• O valor do desconto: R$ {vd}")
print(f"• O valor do final a pagar: R$ {f}")
print(f"• O valor final é menor que 75.00? {l}")
```
