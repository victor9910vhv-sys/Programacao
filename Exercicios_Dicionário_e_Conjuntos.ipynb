DICIONÁRIOS

```python
# 1. Crie um dicionário chamado aluno contendo: nome, idade e curso. Mostre o dicionário.

aluno = {
    "nome": "Victor",
    "idade": 16,
    "curso": "Informática"
}

print(aluno)
```

```python
# 2. Crie um dicionário com nome, idade e curso de um aluno. Mostre cada informação separadamente.

aluno = {
    "nome": "Victor",
    "idade": 16,
    "curso": "Informática"
}

print(aluno["nome"])
print(aluno["idade"])
print(aluno["curso"])
```

```python
# 3. Crie um dicionário produto contendo nome, preço e estoque. Altere o preço e mostre o resultado.

produto = {
    "nome": "Caderno",
    "preço": 20.00,
    "estoque": 10
}

produto["preço"] = 25.00

print(produto)
```

```python
# 4. Crie um dicionário aluno contendo nome e idade. Depois, adicione a informação curso.

aluno = {
    "nome": "Victor",
    "idade": 16
}

aluno["curso"] = "Informática"

print(aluno)
```

```python
# 5. Crie um dicionário chamado filme contendo: título, ano, gênero e nota. Depois, mostre cada informação separadamente.

filme = {
    "título": "Homem-Aranha",
    "ano": 2021,
    "gênero": "Ação",
    "nota": 9
}

print(filme["título"])
print(filme["ano"])
print(filme["gênero"])
print(filme["nota"])
```

CONJUNTOS

```python
# 1. Crie um conjunto chamado numeros contendo os números de 1 a 5. Mostre o conjunto.

numeros = {1, 2, 3, 4, 5}

print(numeros)
```

```python
# 2. Crie numeros = {1, 2, 2, 3, 3, 3, 4, 5}. Mostre o conjunto e observe o que acontece com os números repetidos.

numeros = {1, 2, 2, 3, 3, 3, 4, 5}

print(numeros)
```

```python
# 3. Crie frutas = {"maçã", "banana", "laranja"}. Adicione "morango" usando add().

frutas = {"maçã", "banana", "laranja"}

frutas.add("morango")

print(frutas)
```

```python
# 4. Crie cores = {"vermelho", "azul", "verde", "amarelo"}. Remova "verde" usando remove().

cores = {"vermelho", "azul", "verde", "amarelo"}

cores.remove("verde")

print(cores)
```

```python
# 5. Crie a lista de nomes. Transforme a lista em um conjunto para eliminar os nomes repetidos.

nomes = ["Yris", "Luna", "Luna", "Maitê", "Yris", "Ravi", "Gael", "Theo", "Theo"]

nomes = set(nomes)

print(nomes)
```
